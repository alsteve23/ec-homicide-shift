```python
import pandas as pd
from pathlib import Path
```

```python

ROOT = Path.cwd().parent
RAW = ROOT / "data" / "raw"
df = pd.read_excel(RAW / "mdi_homicidiosintencionales_pm_2014_2025.xlsx", sheet_name=1)

df.head()
print(df.shape)
print(df.columns.tolist())
df["fecha_infraccion"] = pd.to_datetime(df["fecha_infraccion"], errors="coerce")
print(df["fecha_infraccion"].min(), df["fecha_infraccion"].max())
print(df.groupby(df["fecha_infraccion"].dt.year).size())
print(df["codigo_canton"].nunique())
print((df.isna().mean()*100).round(1).sort_values(ascending=False))
for c in ["tipo_muerte", "tipo_arma", "presunta_motivacion", "medida_edad"]:
    if c in df.columns:
        print(df[c].value_counts(dropna=False).head(10))
```

```python
import numpy as np
df = df.replace("SIN_DATO", np.nan)
print((df.isna().mean()*100).round(1).sort_values(ascending=False).head(12))

for c in ["arma", "presun_motiva_observada", "probable_causa_motivada", "area_hecho"]:
    print(df[c].value_counts(dropna=False).head(12), "\n")

print(df["codigo_canton"].dtype, df["codigo_canton"].head())
print(df[["coordenada_x", "coordenada_y"]].describe())
```

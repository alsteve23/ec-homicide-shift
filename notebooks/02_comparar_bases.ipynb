```python
from pathlib import Path
import pandas as pd
```

```python
ROOT = Path.cwd().parent
RAW = ROOT / "data" / "raw"

h = pd.read_excel(RAW / "mdi_homicidiosintencionales_pm_2014_2025.xlsx", sheet_name=1)
n = pd.read_excel(RAW / "mdi_homicidiosintencionales_pm_2026_enero_agosto.xlsx", sheet_name=1)
```

```python
# 1. Same columns, same order?
print("Same columns:", list(h.columns) == list(n.columns))
print("Differences:", set(h.columns) ^ set(n.columns))

# 2. Same types?
tipos = pd.concat([h.dtypes, n.dtypes], axis=1, keys=["2014_2025", "2026"])
print(tipos[tipos["2014_2025"] != tipos["2026"]])

# 3. 2026: date range and total
n["fecha_infraccion"] = pd.to_datetime(n["fecha_infraccion"], errors="coerce")
print(n["fecha_infraccion"].min(), n["fecha_infraccion"].max(), len(n))
print(n.groupby(n["fecha_infraccion"].dt.month).size())

# 4. New or missing categories in the variables we will use
cols = ["tipo_muerte", "area_hecho", "arma", "presun_motiva_observada",
        "sexo", "medida_edad", "codigo_canton"]
for c in cols:
    a, b = set(h[c].dropna().astype(str)), set(n[c].dropna().astype(str))
    print(f"\n{c}\n  only in 2026: {b - a}\n  only in 2014-2025: {len(a - b)} categories")
```
